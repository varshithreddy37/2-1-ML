```python
import pandas as pd
import numpy as np
```

```python
# Load the original CKD dataset

file_path = r"C:\Users\Arnav\OneDrive\Desktop\KLH\2year\Machine Learning\Hybrid_ML_CKD_Prediction\dataset\raw\chronic_kidney_disease.csv"

df = pd.read_csv(file_path)

print("Raw dataset loaded successfully.")
print("Shape:", df.shape)
```

```python
#  copy for preprocessing

df_processed = df.copy()

print("Working copy created successfully.")
```

```python
# Check missing values before preprocessing

missing_before = df_processed.isnull().sum()

missing_before = missing_before[missing_before > 0]

print("Missing values before preprocessing:")
print(missing_before)

print("\nTotal missing cells before preprocessing:",
      df_processed.isnull().sum().sum())
```

```python
# Define numerical and categorical columns

numerical_columns = [
    "age",
    "bp",
    "sg",
    "al",
    "su",
    "bgr",
    "bu",
    "sc",
    "sod",
    "pot",
    "hemo",
    "pcv",
    "wbcc",
    "rbcc"
]

categorical_columns = [
    "rbc",
    "pc",
    "pcc",
    "ba",
    "htn",
    "dm",
    "cad",
    "appet",
    "pe",
    "ane"
]

target_column = "class"

print("Numerical columns:", len(numerical_columns))
print("Categorical columns:", len(categorical_columns))
print("Target column:", target_column)
```

```python
# Calculate median values for numerical columns

numerical_medians = df_processed[numerical_columns].median()

print("Median values of numerical columns:")
print(numerical_medians)
```

```python
# categorical columns using the mode

for column in categorical_columns:
    mode_value = df_processed[column].mode()[0]
    df_processed[column] = df_processed[column].fillna(mode_value)

print("Categorical missing values filled using mode.")
```

```python
# Check missing values after imputation

missing_after = df_processed.isnull().sum()

print("Missing values after preprocessing:")
print(missing_after[missing_after > 0])

print("\nTotal missing cells after preprocessing:",
      df_processed.isnull().sum().sum())
```

```python
# Verify categorical values after missing-value treatment

print("=" * 60)
print("CATEGORICAL VALUES AFTER IMPUTATION")
print("=" * 60)

for column in categorical_columns:
    print(f"\n{column}:")
    print(df_processed[column].value_counts())
```

```python
# Final missing-value verification

total_missing_after = df_processed.isnull().sum().sum()

print("=" * 60)
print("FINAL MISSING-VALUE VERIFICATION")
print("=" * 60)

print("Total missing cells before preprocessing:", 1012)
print("Total missing cells after preprocessing:", total_missing_after)

if total_missing_after == 0:
    print("\nSUCCESS: All missing values have been handled.")
else:
    print("\nWARNING: Some missing values are still present.")
```

```python
# Identify remaining missing values after preprocessing

remaining_missing = df_processed.isnull().sum()

remaining_missing = remaining_missing[remaining_missing > 0]

print("=" * 60)
print("REMAINING MISSING VALUES")
print("=" * 60)

print(remaining_missing)

print("\nTotal remaining missing cells:",
      df_processed.isnull().sum().sum())
```

```python
# Check for question marks and other possible missing-value markers

print("=" * 60)
print("CHECKING FOR '?' VALUES")
print("=" * 60)

for column in df_processed.columns:
    question_marks = (df_processed[column].astype(str).str.strip() == "?").sum()

    if question_marks > 0:
        print(f"{column}: {question_marks} '?' values")
```

```python
# Check exactly which columns still contain missing values

remaining_missing = df_processed.isnull().sum()

print("=" * 60)
print("REMAINING MISSING VALUES AFTER PREPROCESSING")
print("=" * 60)

print(remaining_missing[remaining_missing > 0])

print("\nTotal remaining missing cells:",
      remaining_missing.sum())
```

```python
# Check for common missing-value markers

missing_markers = ["?", "", "NA", "N/A", "null", "None"]

print("=" * 60)
print("CHECKING FOR NON-NaN MISSING VALUE MARKERS")
print("=" * 60)

for column in df_processed.columns:
    for marker in missing_markers:
        count = (
            df_processed[column]
            .astype(str)
            .str.strip()
            .str.lower()
            .eq(marker.lower())
            .sum()
        )

        if count > 0:
            print(f"{column}: '{marker}' -> {count}")
```

```python
# Fill missing values in numerical columns using the median

for column in numerical_columns:
    median_value = df_processed[column].median()
    df_processed[column] = df_processed[column].fillna(median_value)

print("Numerical missing values filled using median.")
```

```python
# Verify numerical missing values after median imputation

print("=" * 60)
print("NUMERICAL MISSING VALUES AFTER IMPUTATION")
print("=" * 60)

for column in numerical_columns:
    missing_count = df_processed[column].isnull().sum()
    print(f"{column}: {missing_count}")

print("\nTotal numerical missing cells:",
      df_processed[numerical_columns].isnull().sum().sum())
```

```python
# Final missing-value verification

total_missing_after = df_processed.isnull().sum().sum()

print("=" * 60)
print("FINAL MISSING-VALUE VERIFICATION")
print("=" * 60)

print("Total missing cells before preprocessing: 1012")
print("Total missing cells after preprocessing:", total_missing_after)

if total_missing_after == 0:
    print("\nSUCCESS: All missing values have been handled.")
else:
    print("\nWARNING: Missing values are still present.")
```
