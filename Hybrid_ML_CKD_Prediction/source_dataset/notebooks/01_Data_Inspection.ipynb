```python
import pandas as pd
```

```python
# Load the original CKD dataset
file_path = r"C:\Users\Arnav\OneDrive\Desktop\KLH\2year\Machine Learning\Hybrid_ML_CKD_Prediction\dataset\raw\chronic_kidney_disease.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
```

```python
print("Dataset Shape:", df.shape)
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python
# Count missing values in each column
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_report = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage.round(2)
})

missing_report = missing_report[
    missing_report["Missing Values"] > 0
]

print(missing_report)
```

```python
total_missing = df.isnull().sum().sum()

print("Total Missing Cells:", total_missing)
```

```python
# Classify columns into numerical and categorical fields

numerical_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_columns = df.select_dtypes(
    include=["object"]
).columns.tolist()

print("=" * 60)
print("NUMERICAL COLUMNS")
print("=" * 60)

for column in numerical_columns:
    print(column)

print("\nTotal Numerical Columns:", len(numerical_columns))


print("\n" + "=" * 60)
print("CATEGORICAL COLUMNS")
print("=" * 60)

for column in categorical_columns:
    print(column)

print("\nTotal Categorical Columns:", len(categorical_columns))
```

```python
# Display the data type of every column

data_type_report = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values
})

print("=" * 60)
print("DATA TYPE OF EVERY FIELD")
print("=" * 60)

print(data_type_report.to_string(index=False))
```

```python
# Create a complete field classification table

classification = []

for column in df.columns:
    if column in numerical_columns:
        data_category = "Numerical"
    else:
        data_category = "Categorical"

    classification.append({
        "Field": column,
        "Data Type": str(df[column].dtype),
        "Category": data_category,
        "Missing Values": int(df[column].isnull().sum()),
        "Missing Percentage": round(
            df[column].isnull().mean() * 100, 2
        )
    })

classification_report = pd.DataFrame(classification)

print(classification_report.to_string(index=False))
```

```python
# Display unique values in each categorical field

print("=" * 60)
print("UNIQUE VALUES IN CATEGORICAL FIELDS")
print("=" * 60)

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].dropna().unique())
```

```python
# Display frequency of each category

print("=" * 60)
print("CATEGORY FREQUENCIES")
print("=" * 60)

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].value_counts(dropna=False))
```

```python
# Statistical summary of numerical fields

print("=" * 60)
print("NUMERICAL FIELD STATISTICS")
print("=" * 60)

numerical_summary = df[numerical_columns].describe().T

numerical_summary["median"] = df[numerical_columns].median()

numerical_summary = numerical_summary[
    [
        "count",
        "mean",
        "median",
        "std",
        "min",
        "25%",
        "50%",
        "75%",
        "max"
    ]
]

print(numerical_summary.round(2))
```

```python
# Check minimum and maximum values of numerical fields

print("=" * 60)
print("MINIMUM AND MAXIMUM VALUES")
print("=" * 60)

min_max_report = pd.DataFrame({
    "Field": numerical_columns,
    "Minimum": [df[column].min() for column in numerical_columns],
    "Maximum": [df[column].max() for column in numerical_columns]
})

print(min_max_report.to_string(index=False))
```

```python
# Display unusually low values for selected clinical fields

print("=" * 60)
print("LOWEST VALUES OF NUMERICAL FIELDS")
print("=" * 60)

for column in numerical_columns:
    print(f"\n{column}:")
    print(df[column].dropna().sort_values().head(10).to_list())
```

```python
# Investigate potentially unusual numerical values

print("=" * 60)
print("POTENTIAL DATA-QUALITY VALUES")
print("=" * 60)

print("\n1. Sodium (sod) values below 100:")
print(df.loc[df["sod"] < 100, ["sod", "class"]].to_string(index=False))

print("\n2. Potassium (pot) values above 10:")
print(df.loc[df["pot"] > 10, ["pot", "class"]].to_string(index=False))

print("\n3. Serum Creatinine (sc) values above 10:")
print(df.loc[df["sc"] > 10, ["sc", "class"]].to_string(index=False))

print("\n4. Blood Urea (bu) values above 150:")
print(df.loc[df["bu"] > 150, ["bu", "class"]].to_string(index=False))

print("\n5. Blood Glucose (bgr) values above 300:")
print(df.loc[df["bgr"] > 300, ["bgr", "class"]].to_string(index=False))
```
