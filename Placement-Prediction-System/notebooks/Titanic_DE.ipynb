```python
# ============================================================
# BOOSTING BENCHMARK - PLACEMENT PREDICTION SYSTEM
# COMPLETE VS CODE VERSION
# ============================================================

import sys
import subprocess
import warnings

warnings.filterwarnings("ignore")

# ============================================================
# 0. INSTALL REQUIRED PACKAGES IF MISSING
# ============================================================

def install_package(package_name):
    try:
        __import__(package_name)
        print(f"{package_name} is already installed.")
        return True
    except ImportError:
        print(f"Installing {package_name}...")
        try:
            subprocess.check_call([
                sys.executable,
                "-m",
                "pip",
                "install",
                package_name
            ])
            print(f"{package_name} installed successfully.")
            return True
        except Exception as e:
            print(f"Could not install {package_name}: {e}")
            return False


xgboost_available = install_package("xgboost")
lightgbm_available = install_package("lightgbm")
shap_available = install_package("shap")


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import time
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import (
    AdaBoostClassifier,
    GradientBoostingClassifier
)
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

if xgboost_available:
    from xgboost import XGBClassifier

if lightgbm_available:
    from lightgbm import LGBMClassifier

if shap_available:
    import shap


# ============================================================
# 2. DATASET PATH
# ============================================================

DATA_PATH = (
    r"C:\Users\pendi\2-1-ML\PLACEMENT PREDICTION SYSTEM"
    r"\data\placement_predict_50k Dataset_final (1).csv"
)


# ============================================================
# 3. LOAD DATASET
# ============================================================

print("\n" + "=" * 70)
print("LOADING DATASET")
print("=" * 70)

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())


# ============================================================
# 4. CHECK DATASET
# ============================================================

if df.empty:
    raise ValueError(
        "ERROR: Dataset is empty. Check the CSV file."
    )

if "PlacementStatus" not in df.columns:
    raise ValueError(
        "ERROR: PlacementStatus column was not found."
    )

print("\nPlacementStatus column found successfully.")

print("\nOriginal PlacementStatus values:")
print(df["PlacementStatus"].value_counts(dropna=False))

print("\nUnique PlacementStatus values:")
print(df["PlacementStatus"].unique())


# ============================================================
# 5. CLEAN TARGET VARIABLE
# ============================================================

print("\n" + "=" * 70)
print("ENCODING PLACEMENT STATUS")
print("=" * 70)

target_raw = df["PlacementStatus"]

# Check if target is numeric
numeric_target = pd.to_numeric(
    target_raw,
    errors="coerce"
)

# If all non-null values are numeric
if numeric_target.notna().sum() == target_raw.notna().sum():

    df["PlacementStatus"] = numeric_target

else:

    target_clean = (
        target_raw
        .astype(str)
        .str.strip()
        .str.lower()
    )

    target_mapping = {

        "yes": 1,
        "no": 0,

        "placed": 1,
        "not placed": 0,

        "not_placed": 0,
        "notplaced": 0,

        "true": 1,
        "false": 0,

        "1": 1,
        "0": 0,

        "y": 1,
        "n": 0
    }

    df["PlacementStatus"] = target_clean.map(
        target_mapping
    )


# ============================================================
# 6. HANDLE UNKNOWN TARGET VALUES
# ============================================================

if df["PlacementStatus"].isna().any():

    unknown_count = df["PlacementStatus"].isna().sum()

    print(
        f"\nWarning: {unknown_count} target values "
        f"could not be directly converted."
    )

    # Try automatic conversion for remaining values
    remaining = (
        df.loc[
            df["PlacementStatus"].isna(),
            "PlacementStatus"
        ]
    )

    # If conversion failed, inspect original values
    original_remaining = target_raw.loc[
        df["PlacementStatus"].isna()
    ]

    print(
        "Unrecognized original values:",
        original_remaining.unique()
    )


# ============================================================
# 7. REMOVE ROWS WITH INVALID TARGET
# ============================================================

df = df.dropna(
    subset=["PlacementStatus"]
).copy()


if df.empty:
    raise ValueError(
        "ERROR: All rows were removed while processing "
        "PlacementStatus. Check the target values printed above."
    )


df["PlacementStatus"] = (
    df["PlacementStatus"]
    .astype(int)
)


print("\nTarget encoding completed.")

print("\nFinal target distribution:")
print(
    df["PlacementStatus"].value_counts()
)


# ============================================================
# 8. CHECK NUMBER OF CLASSES
# ============================================================

unique_target = df["PlacementStatus"].nunique()

if unique_target != 2:
    raise ValueError(
        f"ERROR: PlacementStatus must contain exactly "
        f"2 classes, but found {unique_target}."
    )


# ============================================================
# 9. DEFINE FEATURES AND TARGET
# ============================================================

TARGET = "PlacementStatus"

drop_columns = [
    TARGET
]

# StudentID is only an identifier
if "StudentID" in df.columns:
    drop_columns.append("StudentID")

# Salary Package is normally available AFTER placement.
# Keeping it would cause data leakage.
if "Salary Package" in df.columns:
    drop_columns.append("Salary Package")

X = df.drop(
    columns=drop_columns
)

y = df[TARGET]


print("\n" + "=" * 70)
print("FEATURE INFORMATION")
print("=" * 70)

print("Number of samples:", len(X))
print("Number of features:", X.shape[1])

print("\nFeatures:")
print(X.columns.tolist())


# ============================================================
# 10. HANDLE CATEGORICAL FEATURES
# ============================================================

print("\n" + "=" * 70)
print("ENCODING FEATURES")
print("=" * 70)

for column in X.columns:

    # Object / categorical column
    if X[column].dtype == "object":

        values = (
            X[column]
            .fillna("")
            .astype(str)
            .str.strip()
            .str.lower()
        )

        unique_values = set(
            values[values != ""].unique()
        )

        # Yes / No
        if unique_values.issubset(
            {"yes", "no"}
        ):

            X[column] = values.map({
                "no": 0,
                "yes": 1
            })

        # Other categorical variables
        else:

            X[column] = (
                X[column]
                .astype("category")
                .cat.codes
            )


# ============================================================
# 11. CONVERT EVERYTHING TO NUMERIC
# ============================================================

for column in X.columns:

    X[column] = pd.to_numeric(
        X[column],
        errors="coerce"
    )


# ============================================================
# 12. HANDLE INFINITE VALUES
# ============================================================

X = X.replace(
    [np.inf, -np.inf],
    np.nan
)


# ============================================================
# 13. HANDLE MISSING VALUES
# ============================================================

for column in X.columns:

    if X[column].isna().any():

        median_value = X[column].median()

        if pd.isna(median_value):
            median_value = 0

        X[column] = X[column].fillna(
            median_value
        )


# ============================================================
# 14. FINAL DATA CHECK
# ============================================================

print("\n" + "=" * 70)
print("FINAL DATA CHECK")
print("=" * 70)

print("Samples:", len(X))
print("Features:", X.shape[1])

print("\nMissing values:")
print(X.isna().sum().sum())

print("\nTarget distribution:")
print(y.value_counts())


# ============================================================
# 15. TRAIN / VALIDATION SPLIT
# ============================================================

X_train, X_val, y_train, y_val = train_test_split(

    X,
    y,

    test_size=0.30,

    random_state=42,

    stratify=y
)


print("\n" + "=" * 70)
print("TRAIN / VALIDATION SPLIT")
print("=" * 70)

print("Training samples:", len(X_train))
print("Validation samples:", len(X_val))


# ============================================================
# 16. CREATE MODELS
# ============================================================

models = {}


# ---------------- ADA BOOST ----------------

models["AdaBoost"] = AdaBoostClassifier(
    n_estimators=100,
    learning_rate=0.1,
    random_state=42
)


# ---------------- GRADIENT BOOSTING ----------------

models["Gradient Boosting"] = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)


# ---------------- XGBOOST ----------------

if xgboost_available:

    models["XGBoost"] = XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        random_state=42,
        eval_metric="logloss",
        n_jobs=-1
    )


# ---------------- LIGHTGBM ----------------

if lightgbm_available:

    models["LightGBM"] = LGBMClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        random_state=42,
        verbosity=-1,
        n_jobs=-1
    )


# ============================================================
# 17. TRAIN BOOSTING MODELS
# ============================================================

print("\n" + "=" * 70)
print("BOOSTING BENCHMARK")
print("=" * 70)

results = []

trained_models = {}


for name, model in models.items():

    print("\nTraining:", name)

    start_time = time.perf_counter()

    model.fit(
        X_train,
        y_train
    )

    end_time = time.perf_counter()

    training_time = (
        end_time - start_time
    )

    trained_models[name] = model

    # Training prediction
    train_pred = model.predict(
        X_train
    )

    # Validation prediction
    val_pred = model.predict(
        X_val
    )

    # Accuracy
    train_accuracy = accuracy_score(
        y_train,
        train_pred
    )

    val_accuracy = accuracy_score(
        y_val,
        val_pred
    )

    results.append({

        "Model": name,

        "Train Accuracy":
            train_accuracy,

        "Validation Accuracy":
            val_accuracy,

        "Training Time (sec)":
            training_time
    })

    print(
        "Train Accuracy:",
        round(train_accuracy * 100, 2),
        "%"
    )

    print(
        "Validation Accuracy:",
        round(val_accuracy * 100, 2),
        "%"
    )

    print(
        "Training Time:",
        round(training_time, 4),
        "seconds"
    )


# ============================================================
# 18. RESULTS DATAFRAME
# ============================================================

results_df = pd.DataFrame(
    results
)


# ============================================================
# 19. SORT RESULTS
# ============================================================

results_df = results_df.sort_values(
    by="Validation Accuracy",
    ascending=False
).reset_index(drop=True)


# ============================================================
# 20. FINAL RESULTS
# ============================================================

print("\n" + "=" * 70)
print("FINAL BOOSTING BENCHMARK RESULTS")
print("=" * 70)

print(
    results_df.to_string(
        index=False
    )
)


# ============================================================
# 21. BEST MODEL
# ============================================================

best_model_name = results_df.iloc[0]["Model"]

best_accuracy = (
    results_df.iloc[0]["Validation Accuracy"]
)


print("\n" + "=" * 70)
print("BEST MODEL")
print("=" * 70)

print(
    "Best Model:",
    best_model_name
)

print(
    "Validation Accuracy:",
    round(
        best_accuracy * 100,
        2
    ),
    "%"
)


# ============================================================
# 22. CLASSIFICATION REPORT
# ============================================================

best_model = trained_models[
    best_model_name
]

best_predictions = best_model.predict(
    X_val
)

print("\n" + "=" * 70)
print("CLASSIFICATION REPORT - BEST MODEL")
print("=" * 70)

print(
    classification_report(
        y_val,
        best_predictions
    )
)


# ============================================================
# 23. CONFUSION MATRIX
# ============================================================

print("\n" + "=" * 70)
print("CONFUSION MATRIX")
print("=" * 70)

cm = confusion_matrix(
    y_val,
    best_predictions
)

print(cm)


# ============================================================
# 24. SAVE BENCHMARK RESULTS
# ============================================================

OUTPUT_PATH = (
    r"C:\Users\pendi\2-1-ML\PLACEMENT PREDICTION SYSTEM"
    r"\data\boosting_benchmark_results.csv"
)

results_df.to_csv(
    OUTPUT_PATH,
    index=False
)

print("\nBenchmark results saved to:")
print(OUTPUT_PATH)


# ============================================================
# 25. XGBOOST SHAP ANALYSIS
# ============================================================

if (
    shap_available
    and xgboost_available
    and "XGBoost" in trained_models
):

    print("\n" + "=" * 70)
    print("XGBOOST SHAP ANALYSIS")
    print("=" * 70)

    try:

        xgb_model = trained_models["XGBoost"]

        # Use a smaller sample for faster SHAP calculation
        shap_data = X_val.copy()

        if len(shap_data) > 2000:
            shap_data = shap_data.sample(
                2000,
                random_state=42
            )

        explainer = shap.TreeExplainer(
            xgb_model
        )

        shap_values = explainer.shap_values(
            shap_data
        )

        plt.figure(
            figsize=(10, 6)
        )

        shap.summary_plot(
            shap_values,
            shap_data,
            plot_type="dot",
            show=False
        )

        plt.title(
            "XGBoost SHAP Beeswarm Plot"
        )

        plt.tight_layout()

        plt.show()

        print(
            "SHAP analysis completed successfully!"
        )

    except Exception as e:

        print(
            "\nSHAP analysis could not be generated."
        )

        print(
            "Reason:",
            e
        )

else:

    print(
        "\nSHAP is not available."
    )

    print(
        "The boosting benchmark was completed without SHAP."
    )


# ============================================================
# 26. V1 VS V2 ACCURACY CHART
# ============================================================

plt.figure(
    figsize=(7, 5)
)

plt.bar(
    results_df["Model"],
    results_df["Validation Accuracy"]
)

plt.ylabel(
    "Validation Accuracy"
)

plt.xlabel(
    "Boosting Model"
)

plt.title(
    "Boosting Model Validation Accuracy"
)

plt.ylim(
    0,
    1
)

plt.xticks(
    rotation=20
)

for i, value in enumerate(
    results_df["Validation Accuracy"]
):

    plt.text(
        i,
        value + 0.02,
        f"{value:.3f}",
        ha="center"
    )

plt.tight_layout()

plt.show()


# ============================================================
# 27. COMPLETED
# ============================================================

print("\n" + "=" * 70)
print("BOOSTING BENCHMARK COMPLETED SUCCESSFULLY")
print("=" * 70)

print(
    "Best Model:",
    best_model_name
)

print(
    "Best Validation Accuracy:",
    round(
        best_accuracy * 100,
        2
    ),
    "%"
)

print(
    "Results file:",
    OUTPUT_PATH
)
```
